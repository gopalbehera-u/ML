```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
```

```python
dataset=pd.read_csv(r"C:\Users\beher\OneDrive\Desktop\ML\classification\logit classification.csv")
```

```python
dataset.head()
```

```python
x=dataset.drop(['User ID','Gender','Purchased'],axis=1)
```

```python
y=dataset['Purchased']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
sc=StandardScaler()
x_train_scaled=sc.fit_transform(x_train)
x_test_scled=sc.transform(x_test)
```

# train model without scaling 

```python
dt=DecisionTreeClassifier(criterion='log_loss',max_depth=4,random_state=0,splitter='best')
dt.fit(x_train,y_train)
y_pred=dt.predict(x_test)
```

```python
ac=accuracy_score(y_test,y_pred)
ac
```

```python
cm=confusion_matrix(y_test,y_pred)
print(cm)
```

```python
cr=classification_report(y_test,y_pred)
print(cr)
```
