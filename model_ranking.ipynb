```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor
import warnings
warnings.filterwarnings('ignore')
```

# Linear reg

```python
dataset=pd.read_csv("emp_sal (1).csv")
```

```python
dataset
```

```python
x=dataset[['Level']].values
y=dataset[['Salary']].values
```

```python
lin_reg=LinearRegression()
lin_reg.fit(x,y)
lin_pred=lin_reg.predict([[6.5]])

lin_pred
```

# polynomia reg


```python
poly_reg=PolynomialFeatures(degree=5)
x_poly=poly_reg.fit_transform(x)

lin_reg_2=LinearRegression()
lin_reg_2.fit(x_poly,y)

lin_poly_pred=lin_reg_2.predict(poly_reg.transform([[6.5]]))
lin_poly_pred
```

# SVR

```python
svr_reg=SVR(kernel='poly',degree=4,C=10.0,gamma='auto')
svr_reg.fit(x,y)
svr_pred=svr_reg.predict([[6.5]])
svr_pred
```

# knn

```python
knn_reg=KNeighborsRegressor(weights='distance',p=1)
knn_reg.fit(x,y)
knn_pred=knn_reg.predict([[6.5]])
knn_pred
```

# Desicion tree


```python
dt_reg=DecisionTreeRegressor(criterion='poisson',max_depth=5,min_samples_split=3)
dt_reg.fit(x,y)
dt_pred=dt_reg.predict([[6.5]])
dt_pred
```

# random forest

```python
rf_reg=RandomForestRegressor(criterion='poisson',random_state=0,n_estimators=20,min_samples_split=3)
rf_reg.fit(x,y)
rf_pred=rf_reg.predict([[6.5]])
rf_pred
```
